```sql
CREATE CATALOG IF NOT EXISTS tpch_sales
  COMMENT 'TPCH project catalog (DEV)';
```

```sql
CREATE SCHEMA IF NOT EXISTS tpch_sales.bronze
    COMMENT "bronze zone for raw tables with metadata";
CREATE SCHEMA IF NOT EXISTS tpch_sales.silver
    COMMENT "silver zone for transformed data";
CREATE SCHEMA IF NOT EXISTS tpch_sales.gold
    COMMENT "gold zone for final tables";
```

```sql
CREATE VOLUME IF NOT EXISTS tpch_sales.bronze.raw;
CREATE VOLUME IF NOT EXISTS tpch_sales.bronze.pipeline_state;
```
